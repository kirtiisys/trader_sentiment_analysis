```python
import pandas as pd
import numpy as np

# Load datasets
sentiment = pd.read_csv("fear_greed_index.csv")
trades = pd.read_csv("historical_data.csv")

# shape of dataset
print("Sentiment Shape:", sentiment.shape)
print("Trades Shape:", trades.shape)

# Missing values
print("Missing Values:", sentiment.isnull().sum())
print("Missing Values:", trades.isnull().sum())

# Duplicates
print("Duplicates in sentiment:", sentiment.duplicated().sum())
print("Duplicates in trades:", trades.duplicated().sum())
```

```python
# Rename columns for simplicity
sentiment = sentiment.rename(columns={
    'classification': 'Sentiment',
    'date': 'Date'
})

trades = trades.rename(columns={
    'Closed PnL': 'PnL',
    'Size USD': 'SizeUSD',
    'Size Tokens': 'SizeTokens'
})
```

```python
# Convert timestamp
sentiment['Date'] = pd.to_datetime(sentiment['Date'])

# Convert trade timestamp
trades['Timestamp'] = pd.to_datetime(trades['Timestamp'], unit='ms')

# Extract date align dataset by date
trades['Date'] = trades['Timestamp'].dt.date
sentiment['Date'] = sentiment['Date'].dt.date
```

```python
merged = pd.merge(trades, sentiment, on="Date", how="left")

print("Merged shape:", merged.shape)
print(merged[['Date','Sentiment']].head())
```

```python
# Feature engineering
merged['Win'] = merged['PnL'] > 0

# Daily PnL per trade
daily_pnl = merged.groupby('Date')['PnL'].sum()

# Win rate
win_rate = merged.groupby('Account')['Win'].mean()

# Trades per day
trades_per_day = merged.groupby('Date').size()

# Avg trade size
avg_size = merged.groupby('Account')['SizeUSD'].mean()
```

```python
# Performance metrics by sentiment
performance = merged.groupby('Sentiment').agg({
    'PnL': ['mean', 'sum'],
    'Win': 'mean'
})

performance.columns = ['Avg_PnL', 'Total_PnL', 'Win_Rate']
print(performance)
```

```python
import matplotlib.pyplot as plt
performance['Avg_PnL'].plot(kind='bar', title="Average PnL: Fear vs Greed")
plt.ylabel("PnL")
plt.show()

performance['Win_Rate'].plot(kind='bar', title="Win Rate: Fear vs Greed")
plt.show()
```

```python
#behavoir change based on sentiment
behavior = merged.groupby('Sentiment').agg({
    'SizeUSD': 'mean',
    'PnL': 'count'
}).rename(columns={'PnL': 'Trade_Count'})

print(behavior)
```

```python
import matplotlib.pyplot as plt
# Trade frequency
behavior['Trade_Count'].plot(kind='bar', title="Trade Frequency by Sentiment")
plt.show()

# Position size
behavior['SizeUSD'].plot(kind='bar', title="Avg Trade Size by Sentiment")
plt.show()
```

```python
#long position vs short position
long_short = merged.groupby(['Sentiment','Side']).size().unstack().fillna(0)
print(long_short)

long_short.plot(kind='bar', stacked=True, title="Long vs Short by Sentiment")
plt.show()
```

```python
# Q3 Trader Segmentation
#high vs low leverage using size as proxy
import matplotlib.pyplot as plt
merged['LeverageSegment'] = np.where(
    merged['SizeUSD'] > merged['SizeUSD'].median(),
    'High','Low'
)

lev_perf = merged.groupby('LeverageSegment')['PnL'].mean()
print(lev_perf)

lev_perf.plot(kind='bar', title="PnL by Leverage Segment")
plt.show()
```

```python
# frequent and infrequent traders
import matplotlib.pyplot as plt
trade_counts = merged['Account'].value_counts()
threshold = trade_counts.median()

merged['FreqSegment'] = merged['Account'].map(
    lambda x: 'Frequent' if trade_counts[x] > threshold else 'Infrequent'
)

freq_perf = merged.groupby('FreqSegment')['PnL'].mean()
print(freq_perf)

freq_perf.plot(kind='bar', title="PnL by Trading Frequency")
plt.show()
```

```python
#consistent vs inconsistent traders
import matplotlib.pyplot as plt
consistency = merged.groupby('Account')['PnL'].std().reset_index()

consistency.columns = ['Account', 'PnL_std']

threshold = consistency['PnL_std'].median()

consistency['Type'] = consistency['PnL_std'].apply(
    lambda x: 'Consistent' if x < threshold else 'Inconsistent'
)
merged = merged.merge(consistency[['Account', 'Type']], on='Account', how='left')

cons_perf = merged.groupby('Type')['PnL'].mean()
print(cons_perf)
cons_perf.plot(kind='bar', title="PnL: Consistent vs Inconsistent")
plt.show()
```

```python
#predictive model profitability prediction
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report

merged['Sentiment'] = merged['Sentiment'].astype(str).str.strip()
merged['SentimentEncoded'] = merged['Sentiment'].map({'Fear':0, 'Greed':1})

merged['ProfitLabel'] = (merged['PnL'] > 0).astype(int)

features = merged[['SizeUSD', 'SentimentEncoded']].fillna(0)
target = merged['ProfitLabel']

print("Dataset size:", features.shape)
X_train, X_test, y_train, y_test = train_test_split(features, target, test_size=0.2)
model = RandomForestClassifier(n_estimators=20, max_depth=5)

print("Training started...")
model.fit(X_train, y_train)
print("Training done!")

print("Accuracy:", model.score(X_test, y_test))

print(classification_report(y_test, model.predict(X_test)))
```

```python
#Clustering traders into behavioral archetypes
from sklearn.cluster import KMeans

trader_features = merged.groupby('Account').agg({
    'PnL':'mean',
    'SizeUSD':'mean'
}).fillna(0)

kmeans = KMeans(n_clusters=3, random_state=42)
trader_features['Cluster'] = kmeans.fit_predict(trader_features)

print(trader_features.head())
```

```python
import matplotlib.pyplot as plt

plt.scatter(
    trader_features['SizeUSD'],
    trader_features['PnL'],
    c=trader_features['Cluster']
)
plt.xlabel("Avg Size")
plt.ylabel("Avg PnL")
plt.title("Trader Clusters")
plt.show()
```
